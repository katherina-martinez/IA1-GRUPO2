# Ejercicios Teóricos

1. Defina con sus propias palabras inteligencia natural, inteligencia artificial y agente.

Inteligencia natural: Se refiere a la capacidad biológica (propia de los seres humanos) para percibir, comprender, predecir y manipular un mundo mucho más grande y complejo que el cerebro mismo. Es la facultad de la mente para razonar y actuar de forma efectiva ante diversas situaciones.

Inteligencia artificial: Es un campo de estudio científico y de ingeniería que no solo busca comprender el funcionamiento de la mente, sino que se esfuerza en construir entidades inteligentes (máquinas) capaces de computar cómo actuar de manera eficaz y segura en una amplia variedad de situaciones novedosas.

Agente: Es cualquier entidad (humana, robótica o de software) que tiene la capacidad de percibir su medioambiente a través de sensores y actuar sobre él mediante el uso de actuadores.
