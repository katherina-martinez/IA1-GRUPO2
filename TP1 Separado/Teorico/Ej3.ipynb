3. ¿Un agente es siempre una computadora?

No. El concepto de agente es una herramienta de análisis de sistemas que abarca mucho más que solo computadoras.
Por ejemplo, tenemos:
Agentes humanos: Cuyo cuerpo posee sensores como ojos y oídos, y actuadores como manos y piernas.

Agentes robóticos: Que utilizan cámaras o sensores infrarrojos como sensores y diversos motores como actuadores.

Agentes de software (o softbots): Que operan en entornos virtuales recibiendo entradas de red o teclado y actuando mediante el envío de paquetes o visualización de información.
