5. Defina cada tipo de agente en función de su **estructura** y dé un ejemplo de cada categoría.

La estructura general de un agente se define por la relación: Agente = arquitectura + programa
. Según el diseño de su programa, se clasifican en:

- Agentes reactivos simples: Su
estructura se basa únicamente en la percepción actual, ignorando el historial previo. Utilizan reglas de condición-acción (si ocurre X, haz Y) para seleccionar la respuesta inmediata.
Ejemplo: Una aspiradora que solo detecta si la baldosa actual está sucia para decidir si debe aspirar.

- Agentes reactivos basados en modelos: Incorporan un estado interno que les permite seguir el rastro de aspectos del mundo que no son visibles en el momento actual. Necesitan un modelo del mundo que describa cómo evoluciona el entorno y cómo afectan las acciones del agente a dicho entorno.
Ejemplo: Un termostato inteligente que modela la inercia térmica de una habitación para evitar ciclos cortos de encendido y apagado.

- Agentes basados en objetivos: Su estructura incluye información sobre metas o situaciones deseables. Actúan evaluando el futuro ("¿qué pasará si hago A?") para encontrar secuencias de acciones que les permitan alcanzar sus objetivos mediante procesos de búsqueda y planificación.
Ejemplo: Un agente de navegación para un taxi autónomo que calcula la ruta necesaria para llegar a un destino específico.

- Agentes basados en utilidad: Utilizan una función de utilidad para medir el grado de éxito o "felicidad" en un estado determinado. Esto les permite elegir la acción que maximice la utilidad esperada cuando hay objetivos en conflicto o múltiples caminos para llegar a una meta.
Ejemplo: Un sistema de climatización que equilibra el confort térmico de los usuarios frente al coste económico de la electricidad.

- Agentes que aprenden: Tienen una estructura dividida en cuatro componentes: un elemento de actuación (el agente básico), un crítico (evalúa la actuación), un elemento de aprendizaje (realiza mejoras) y un generador de problemas (propone nuevas experiencias).
Ejemplo: Un programa de ajedrez que, tras jugar varias partidas, analiza sus derrotas y victorias para refinar su estrategia de juego.
