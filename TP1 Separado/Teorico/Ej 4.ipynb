4. Defina Omnisciencia, Aprendizaje y Autonomía.

Omnisciencia: Es la capacidad de saberlo todo o conocer con total certeza el resultado real de una acción antes de ejecutarla. La racionalidad no exige ser omnisciente, ya que depende solo de lo que el agente ha percibido hasta el momento.

Aprendizaje: Es el proceso de modificar los componentes internos del agente para que se comporten de forma más acorde con la información recibida. Permite al agente utilizar sus percepciones no solo para actuar, sino para mejorar su habilidad de actuación futura basándose en la experiencia.

Autonomía: Se da cuando el agente es capaz de actuar basándose en su propia experiencia y aprendizaje, compensando así el conocimiento parcial o incorrecto que pudo haberle proporcionado inicialmente su diseñador. Un agente que solo depende del conocimiento previo de su creador carece de autonomía.
